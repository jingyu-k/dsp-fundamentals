```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 시간축 만들기: 1초 동안 1000번 측정
fs = 1000                       # 샘플링 주파수 (1초에 몇 번 재는가)
t = np.arange(0, 1, 1/fs)       # 0초부터 1초까지 0.001초 간격

# 2. 신호 만들기: 5Hz 사인파 + 20Hz 사인파
signal = np.sin(2*np.pi*5*t) + 2*np.sin(2*np.pi*30*t)
signal = signal + np.random.randn(len(t))

# 3. 푸리에 변환
spectrum = np.abs(np.fft.rfft(signal))      # 각 주파수 성분의 크기
freqs = np.fft.rfftfreq(len(signal), 1/fs)  # 각 성분이 몇 Hz인지

# 4. 그래프 그리기
fig, ax = plt.subplots(2, 1, figsize=(10, 6))
ax[0].plot(t, signal)
ax[0].set_title("Time domain")
ax[0].set_xlabel("Time (s)")
ax[1].plot(freqs, spectrum)
ax[1].set_title("Frequency domain")
ax[1].set_xlabel("Frequency (Hz)")
ax[1].set_xlim(0, 50)
plt.tight_layout()
plt.show()
```
